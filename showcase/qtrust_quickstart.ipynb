## 1. Scan a live TLS endpoint

```python
!crypto-inspector scan example.com -c nist,cnsa -o scan.json
```

```python
import json
r = json.load(open('scan.json'))
print('target:', r['target'], '| findings:', len(r['findings']))
for f in r['findings'][:5]:
    print('-', f.get('algorithm'), f.get('key_size'), f.get('location'))
```

## 2. Same scan from Python

```python
from qtrust_inspector.scanner import scan_host

res = scan_host('example.com')
print('findings:', res.finding_count)
print(res.by_algorithm)
```
